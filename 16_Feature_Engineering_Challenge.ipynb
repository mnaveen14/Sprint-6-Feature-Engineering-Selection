## **5. Create Numerical Features**

**Understand the Concept**

Numerical features are created using existing numerical columns.

**Demonstrate the Concept**

- Create useful calculations.
- Use income, loan, expense, and savings columns.
- Check the new features.

```python
# Create numerical features
df["Loan_to_Income"] = df["Loan_Amount"] / (df["Annual_Income"] * 12 + 1)
df["Expense_to_Income"] = df["Monthly_Expenses"] / (df["Annual_Income"] / 12 + 1)
df["Savings_to_Income"] = df["Savings"] / (df["Annual_Income"] + 1)
df["Loan_Per_Month"] = df["Loan_Amount"] / df["Loan_Tenure_Months"]

# new features
print(df[["Loan_to_Income","Expense_to_Income","Savings_to_Income","Loan_Per_Month"]].head())
```

**Explanation**

- Loan_to_Income - compares loan amount with income.
- Expense_to_Income - compares expenses with income.
- Savings_to_Income - compares savings with income.
- Loan_Per_Month - shows loan amount per month.

## **6. Create Categorical Features**

**Understand the Concept**

Categorical features group numerical values into simple categories.

**Demonstrate the Concept**

- Create credit score groups.
- Create income groups.
- Check the new categories.

```python
# Create Categorical Features
df["Credit_Score_Group"] = pd.cut(df["Credit_Score"],bins=[0, 580, 670, 740, 850],labels=["Poor", "Fair", "Good", "Excellent"])
df["Income_Group"] = pd.qcut(df["Annual_Income"],q=4,labels=["Low", "Medium", "High", "Very_High"])

# new features
print(df[["Credit_Score_Group","Income_Group"]].head())
```

**Explanation**

- Credit_Score_Group - groups customers by credit score.
- Income_Group - groups customers by income.
- pd.cut() - creates fixed ranges.
- pd.qcut() - creates equal-sized groups.

## **7. Create Date/Time Features**

**Understand the Concept**

Date columns can provide useful information for the model.

**Demonstrate the Concept**

- Convert the date column.
- Extract year and month.
- Create a day of week feature.

```python
# Convert date column
df["Application_Date"] = pd.to_datetime(df["Application_Date"])

# Create date features
df["Application_Year"] = df["Application_Date"].dt.year
df["Application_Month"] = df["Application_Date"].dt.month
df["Application_Day"] = df["Application_Date"].dt.day
df["Application_DayOfWeek"] = df["Application_Date"].dt.dayofweek

# Display new features
print(df[["Application_Year","Application_Month","Application_Day","Application_DayOfWeek"]].head())
```

**Explanation**

- Application_Year - shows the application year.
- Application_Month - shows the application month.
- Application_Day - shows the application day.
- Application_DayOfWeek - shows the day of the week.

## **8. Create Aggregation Features**

**Understand the Concept**

Aggregation features give summary information for each customer.

**Demonstrate the Concept**

- Count previous applications.
- Calculate average loan amount.
- Use customer-level information.

```python
# Create Aggregation Features
df = df.sort_values(["Customer_ID", "Application_Date"])
df["Previous_Application_Count"] = (df.groupby("Customer_ID").cumcount())
df["Customer_Avg_Loan"] = df.groupby("Customer_ID")["Loan_Amount"].transform("mean")

# new features
print(df[["Customer_ID","Previous_Application_Count","Customer_Avg_Loan"]].head())
```

**Explanation**

- Previous_Application_Count - shows previous applications.
- Customer_Avg_Loan - shows the customer's average loan amount.
- groupby() - groups data by customer.
- cumcount() - counts previous records.

## **9. Create Interaction Features**

**Understand the Concept**

Interaction features combine two or more features.

**Demonstrate the Concept**

- Combine credit score and income.
- Combine loan amount and credit score.
- Create useful relationships between features.

```python
# Create interaction features
df["Income_Credit_Score"] = (df["Annual_Income"] * df["Credit_Score"])
df["Loan_Credit_Ratio"] = (df["Loan_Amount"] / (df["Credit_Score"] + 1))
df["Income_per_Dependent"] = (df["Annual_Income"] / (df["Dependents"] + 1))

# new features
print(df[["Income_Credit_Score","Loan_Credit_Ratio","Income_per_Dependent"]].head())
```

**Explanation**

- Income_Credit_Score - combines income and credit score.
- Loan_Credit_Ratio - compares loan amount with credit score.
- Income_per_Dependent - shows income for each dependent.

## **11. Check for Feature Leakage**

**Understand the Concept**

Feature leakage happens when a feature uses information from the target or future data.

**Demonstrate the Concept**

- Check the features for target information.
- Check correlations with the target.
- Identify possible leakage.

```python
# Check for Feature Leakage - correlations
correlation = df.select_dtypes(include="number").corr()["Loan_Defaulted"]
print("Correlation with Target:")
print(correlation.sort_values(ascending=False))
```

**Explanation**

- corr() - checks the relationship between features.
- Loan_Defaulted - is the target.
- Very high correlation should be checked for possible leakage.

## **13. Perform Feature Selection**

**Understand the Concept**

Feature selection helps us keep the most useful features.

**Demonstrate the Concept**

- Select numerical features.
- Use mutual information.
- Select the top features.

```python
from sklearn.feature_selection import SelectKBest, mutual_info_classif

# features and target
X = df.drop(columns=["Loan_Defaulted"])
y = df["Loan_Defaulted"]

# numerical features
X_numeric = X.select_dtypes(include="number").fillna(0)

# Select 10 features
selector = SelectKBest(score_func=mutual_info_classif,k=min(10, X_numeric.shape[1]))
X_selected = selector.fit_transform(X_numeric, y)
selected_features = X_numeric.columns[selector.get_support()]
print("Selected Features:")
print(selected_features.tolist())
```

**Explanation**

- SelectKBest() - selects the best features.
- mutual_info_classif() - checks feature relationship with the target.
- get_support() - returns the selected features.

## **4. What is Feature Engineering?**

**Understand the Concept**

* Feature Engineering means creating or changing features to make them more useful for a Machine Learning model.
* We can create new features from existing columns.
* We can also transform existing features into a better form.
* The main goal is to give the model useful information to learn from.

**Demonstrate the Concept**

**Example:**

* In the E-Commerce dataset, we can create `Average_Spending_Per_Order` using `Total_Spending` and `Order_Count`.
* This gives information about how much a customer spends on each order.

**AI/ML Usage:**

* Feature Engineering helps the model understand important patterns in the data.
* Well-designed features can improve prediction performance.


```python
# Create a new feature: Average Spending Per Order

df["Average_Spending_Per_Order"] = round((df["Total_Spending"] / df["Order_Count"]),2)
df[["Total_Spending", "Order_Count", "Average_Spending_Per_Order"]].head()
```

**Explanation**

* `Average_Spending_Per_Order` is created using `Total_Spending` and `Order_Count`.
* It shows how much a customer spends on average for each order.

**Justify Your Feature**

* I created this feature because it gives useful information about a customer's average spending behavior.

## **5. Why Feature Engineering?**

**Understand the Concept**

* Raw data may not contain all the information needed by a Machine Learning model.
* Feature Engineering helps us create useful information from existing data.
* Good features can make patterns easier for the model to learn.
* It can improve model performance and help us understand the data better.



## **6. Importance of Domain Knowledge**

**Understand the Concept**

* Domain knowledge means understanding the business or field related to the data.
* It helps us create features that have real-world meaning.
* A feature may be useful when we understand how the business works.
* Good domain knowledge helps us avoid creating meaningless features.

**Demonstrate the Concept**

**Example:**

* In an E-Commerce business, customers who have not purchased for a long time may be more likely to churn.
* We can create a `Days_Since_Last_Purchase` feature using the customer's last purchase date.

**AI/ML Usage:**

* Domain knowledge helps us create features that represent real customer behavior.
* These meaningful features can help the model make better predictions.

```python
# Convert Last_Purchase_Date to datetime
df["Last_Purchase_Date"] = pd.to_datetime(df["Last_Purchase_Date"])

# Create a reference date
reference_date = pd.Timestamp("2026-01-01")

# Create Days Since Last Purchase feature
df["Days_Since_Last_Purchase"] = (reference_date - df["Last_Purchase_Date"]).dt.days
df[["Customer_ID", "Last_Purchase_Date","Days_Since_Last_Purchase", "Churn"]].head()
```

**Explanation**

* `Days_Since_Last_Purchase` shows how many days have passed since the customer's last purchase.
* This feature represents customer recency, which is important in E-Commerce.
* A higher value may indicate that a customer has become less active.

**Justify Your Feature**

* I created this feature because recent customer activity can provide useful information for predicting churn.

## **9. Feature Creation**

**Understand the Concept**

* Feature Creation means creating a new feature from existing data.
* New features can combine one or more columns.
* The new feature should provide useful information that is not directly available.
* Feature Creation often uses formulas or business rules.

**Demonstrate the Concept**

**Example:**

* We can combine `Order_Count` and `Total_Spending` to create `Average_Spending_Per_Order`.

**AI/ML Usage:**

* New features can give the model additional information about customer behavior.

```python
# Create a new feature 

df["Average_Spending_Per_Order"] = (df["Total_Spending"] / df["Order_Count"])

df[["Order_Count", "Total_Spending", "Average_Spending_Per_Order"]].head()
```

**Explanation**

* The new feature calculates the average spending for each order.
* It combines information from two existing columns.
* This gives a clearer view of customer spending behavior.

**Justify Your Feature**

* I created this feature because average spending can help understand customer purchasing behavior.

## **10. Feature Transformation**

**Understand the Concept**

* Feature Transformation means changing the form or scale of an existing feature.
* Transformation can make data easier for a model to work with.
* Common transformations include scaling, logarithmic transformation, and normalization.
* The choice of transformation depends on the data and Machine Learning algorithm.

**Demonstrate the Concept**

**Example:**

* `Total_Spending` may contain values with a large range.
* We can use a logarithmic transformation to reduce the effect of very large values.

**AI/ML Usage:**

* Feature transformation can make numerical features more suitable for Machine Learning models.

```python
import numpy as np

# Create a log-transformed spending feature

df["Log_Total_Spending"] = np.log1p(df["Total_Spending"])

df[["Total_Spending", "Log_Total_Spending"]].head()
```

**Explanation**

* `log1p()` applies a logarithmic transformation to `Total_Spending`.
* It reduces the effect of very large spending values.
* The original feature is kept so we can compare both versions.

**Justify Your Feature**

* I created this transformed feature because it can reduce the impact of highly large spending values.

## **12. Feature Extraction**

**Understand the Concept**

* Feature Extraction means extract useful information from existing data.
* It can convert complex or detailed data into simpler features.
* The extracted information can be used as input for a Machine Learning model.
* Dates, text, and other complex data can be sources for feature extraction.

**Demonstrate the Concept**

**Example:**

* From `Last_Purchase_Date`, we can extract the day, month, or day of the week.
* These extracted values can provide information about customer purchasing behavior.

**AI/ML Usage:**

* Feature extraction converts useful information from raw data into features that a model can understand.

```python
# Extract useful information from Last_Purchase_Date

df["Last_Purchase_Month"] = df["Last_Purchase_Date"].dt.month
df["Last_Purchase_Day"] = df["Last_Purchase_Date"].dt.day
df["Last_Purchase_Weekday"] = df["Last_Purchase_Date"].dt.dayofweek

df[["Last_Purchase_Date","Last_Purchase_Month","Last_Purchase_Day","Last_Purchase_Weekday"]].head()
```

**Explanation**

* Month, day, and weekday are extracted from the purchase date.
* These features provide additional information about when customers made their purchases.
* Date and time feature engineering will be covered in detail in Notebook 4.

**Justify Your Feature**

* I extracted these features because purchase timing may provide useful information about customer behavior.

## **13. Feature Leakage**

**Understand the Concept**

* Feature Leakage happens when a feature contains information that should not be available when making a prediction.
* Leakage can make model performance look better than it really is.
* Leakage can happen when future information or target information is included in the features.
* Features should only use information that would be available at prediction time.

**Demonstrate the Concept**

**Example:**

* Suppose we are predicting whether a customer will churn.
* Using a feature created after the customer has already churned can cause leakage.
* For example, `Churn_Reason` would not be available before the churn prediction.

**AI/ML Usage:**

* Preventing leakage helps ensure that the model learns from realistic information.
* Leakage checks should be performed before training the final model.

```python
# Example of a safe feature

df["Days_Since_Last_Purchase"] = ( pd.Timestamp("2026-01-01") - df["Last_Purchase_Date"]).dt.days
df[["Last_Purchase_Date", "Days_Since_Last_Purchase", "Churn"]].head()
```

**Explanation**

* `Days_Since_Last_Purchase` uses the customer's purchase date to calculate recency.
* The feature does not directly use the `Churn` target.
* We must make sure that the reference date represents the prediction time and does not use future information.

**Justify Your Feature**

* I selected this feature because it uses available customer activity information without directly using the target.

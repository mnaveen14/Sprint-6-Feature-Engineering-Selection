# **Notebook 6 – Interaction Features**

```python
import pandas as pd 
df = pd.read_csv("employee_compensation_performance_dataset.csv")
df.info()
```

## **1. What are Interaction Features?**

**Understand the Concept**

* Interaction features combine two or more existing features.
* They help capture relationships between features.
* The relationship may be missed when features are used separately.


## **2. Feature × Feature**

**Understand the Concept**

* Feature × Feature means multiplying two features.
* It combines the values of two features into one new feature.

**Demonstrate the Concept**

**Example:**

* Quantity × Unit_Price = total value of items.


```python
# Feature × Feature
df["Total_Value"] = (df["Quantity"] * df["Unit_Price"])
df[["Quantity","Unit_Price","Total_Value"]].head()
```

**Explanation**

* Quantity is multiplied by Unit_Price.
* The result is stored in Total_Value.

**Justify Your Feature**

* I created this feature to capture the combined effect of quantity and price.

## **3. Feature + Feature**

**Understand the Concept**

* Feature + Feature means adding two features together.
* It combines related values into one new feature.

**Demonstrate the Concept**

**Example:**

* Monthly_Income + Bonus = total compensation.

```python
df["Total_Compensation"] = (df["Monthly_Income"] + df["Bonus"])
df[["Monthly_Income","Bonus","Total_Compensation"]].head()
```

**Explanation**

* Monthly_Income and Bonus are added together.
* The result is stored in Total_Compensation.
* It represents the employee's total compensation.

**Justify Your Feature**

* I created this feature to capture the total compensation received by an employee.

## **4. Feature − Feature**

**Understand the Concept**

* Feature − Feature means subtracting one feature from another.
* It helps identify the difference or gap between two values.

**Demonstrate the Concept**

**Example:**

* Current_Salary − Previous_Salary = salary growth.

```python
df["Salary_Difference"] = (df["Current_Salary"] - df["Previous_Salary"])
df[["Current_Salary","Previous_Salary","Salary_Difference"]].head()
```

**Explanation**

* Previous_Salary is subtracted from Current_Salary.
* The result is stored in Salary_Difference.

**Justify Your Feature**

* I created this feature to measure the change in employee salary.

## **5. Feature / Feature**

**Understand the Concept**

* Feature / Feature means dividing one feature by another.
* It helps compare two values using a ratio.
* Division by zero should be handled carefully.

**Demonstrate the Concept**

**Example:**

* Experience_Years / Age = experience relative to age.

```python
df["Experience_Age_Ratio"] = (df["Experience_Years"] / df["Age"])
df[["Experience_Years","Age","Experience_Age_Ratio"]].head()
```

**Explanation**

* Experience_Years is divided by Age.
* The result is stored in Experience_Age_Ratio.

**Justify Your Feature**

* I created this feature to compare employee experience relative to age.

## **6. Polynomial Features**

**Understand the Concept**

* Polynomial features create powers and combinations of existing features.
* Common examples are square and interaction terms.
* They are useful for capturing non-linear relationships.

**Demonstrate the Concept**

**Example:**

* Age² can capture a non-linear relationship between age and performance.

```python
df["Age_Squared"] = df["Age"] ** 2
df[["Age","Age_Squared"]].head()
```

**Explanation**

* Age is multiplied by itself using the power operator.
* The result is stored in Age_Squared

**Justify Your Feature**

* I created this feature to capture a possible non-linear relationship between age and performance.

## **7. Pairwise Interactions**

**Understand the Concept**

* Pairwise interactions combine two features at a time.
* They help identify relationships between pairs of features.
* Only meaningful combinations should be created.

**Demonstrate the Concept**

**Example:**

* Training_Hours × Projects_Completed can represent the relationship between training and project work.

```python
df["Training_Project_Interaction"] = (df["Training_Hours"] * df["Projects_Completed"])
df[["Training_Hours","Projects_Completed","Training_Project_Interaction"]].head()
```

**Explanation**

* Training_Hours and Projects_Completed are multiplied.
* The result is stored in Training_Project_Interaction.

**Justify Your Feature**

* I created this feature to capture the combined effect of training and project experience.

## **8. Domain-Based Interactions**

**Understand the Concept**

* Domain-based interactions are created using business or domain knowledge.
* They combine features that have a meaningful relationship in a specific domain.
* The interaction should have a clear business meaning.

**Demonstrate the Concept**

**Example:**

* Total_Spending / Number_of_Orders = average spending per order.

```python
df["Average_Spending_Per_Order"] = (df["Total_Spending"] / df["Number_of_Orders"])
df[["Total_Spending","Number_of_Orders","Average_Spending_Per_Order"]].head()
```

**Explanation**

* Total_Spending is divided by Number_of_Orders.
* The result is stored in Average_Spending_Per_Order.
* It represents the average amount spent for each order.

**Justify Your Feature**

* I created this feature to measure the average spending per order.

## **9. When Interaction Features Are Useful**

* They are useful when two features together provide more information than using them separately.
* They can help capture relationships that simple features cannot represent.
* Domain knowledge helps create meaningful interactions.



## **10. When They Create Unnecessary Complexity**

* Creating too many interactions can increase the number of features.
* Some interactions may be redundant or meaningless.
* Too many features can make the model more complex and may increase overfitting.

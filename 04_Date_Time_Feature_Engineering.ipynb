```python
import pandas as pd 
df = pd.read_csv("ecommerce_customer_transactions_dataset.csv")
df.info()
```

## **1. Year**

**Understand the Concept**

* Year means extracting the year from a date.
* It helps us understand when an event happened.

**Demonstrate the Concept**

**Example:**

* From `2025-06-15`, we can extract `2025` as the year.

* The year is extracted and stored in `Transaction_Year`.

**Justify Your Feature**

* I created this feature to identify yearly transaction patterns.

## **10. Days Since Event**

**Understand the Concept**

* Days Since Event measures how many days have passed since an event occurred.
* It converts a date into a useful time difference.

**Demonstrate the Concept**

**Example:**

* If a customer purchased 20 days ago, `Days_Since_Purchase` is `20`.

```python
reference_date = df["Transaction_Date"].max()

df["Days_Since_Transaction"] = (reference_date - df["Transaction_Date"]).dt.days
df[["Transaction_Date", "Days_Since_Transaction"]].sample(3)
```

**Explanation**

* The latest transaction date is used as the reference date.
* The number of days between each transaction and the reference date is calculated.
  
**Justify Your Feature**

* I created this feature to measure how recently each transaction occurred.

## **11. Time Difference**

**Understand the Concept**

* Time Difference measures the gap between two dates or times.
* It helps understand how much time passed between events.

**Demonstrate the Concept**

**Example:**

* If two purchases happened 15 days apart, the time difference is 15 days.


```python
df = df.sort_values(["Customer_ID", "Transaction_Date"]).reset_index(drop=True)

df["Previous_Transaction_Date"] = ( df.groupby("Customer_ID")["Transaction_Date"].shift(1))

df["Days_Between_Transactions"] = (df["Transaction_Date"] - df["Previous_Transaction_Date"]).dt.days
df[["Customer_ID","Transaction_Date","Previous_Transaction_Date","Days_Between_Transactions"]].head(10)
```

**Explanation**

* The previous transaction date is found for each customer.
* The difference between the current and previous transaction is calculated.
* The first transaction of each customer has no previous transaction.

**Justify Your Feature**

* I created this feature to understand how frequently customers make purchases.

## **12. Customer Age**

**Understand the Concept**

* Customer Age represents the customer's age when the transaction happened.
* It can be calculated from the customer's birth date.

**Demonstrate the Concept**

**Example:**

* A customer born in 1998 may be around 27 years old in 2025.


```python
# Create an example date of birth from customer age
df["Date_of_Birth"] = (df["Transaction_Date"]- pd.to_timedelta(df["Customer_Age"] * 365.25, unit="D"))

df["Calculated_Age"] = ((df["Transaction_Date"] - df["Date_of_Birth"]).dt.days / 365.25).round().astype(int)
df[["Customer_Age", "Date_of_Birth", "Calculated_Age"]].head()
```

**Explanation**

* A date of birth is derived from the available customer age for practice.
* Age is then calculated using the transaction date and date of birth.

**Justify Your Feature**

* I created this feature to understand customer behavior based on age.

## **14. Recency**

**Understand the Concept**

* Recency measures how recently a customer made a purchase.
* A smaller recency value means the customer purchased more recently.

**Demonstrate the Concept**

**Example:**

* A customer who purchased 5 days ago has better recency than one who purchased 60 days ago.


```python
customer_last_purchase = (df.groupby("Customer_ID")["Transaction_Date"].max())
reference_date = df["Transaction_Date"].max()

customer_recency = (reference_date - customer_last_purchase).dt.days
df["Customer_Recency_Days"] = (df["Customer_ID"].map(customer_recency))
df[["Customer_ID", "Transaction_Date", "Customer_Recency_Days"]].sample(5)
```

**Explanation**

* The last purchase date is found for each customer.
* The number of days from the reference date to the last purchase is calculated.

**Justify Your Feature**

* I created this feature to measure how recently each customer purchased.

## **15. Frequency**

**Demonstrate the Concept**

## **16. Seasonality**

**Understand the Concept**

* Seasonality means patterns that repeat during specific time periods.
* These patterns can occur monthly, weekly, or yearly.
* E-commerce transactions may increase during certain seasons or months.

**Demonstrate the Concept**

**Example:**

* Sales may increase during festival seasons or holiday months.


```python
# Create a simple season based on transaction month

def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Summer"
    elif month in [6, 7, 8, 9]:
        return "Monsoon"
    else:
        return "Autumn"
```

```python
df["Season"] = df["Transaction_Date"].dt.month.map(get_season)
df[["Transaction_Date", "Season"]].head()
```

```python
# season and calculate total amount
season_sales = (df.groupby("Season")["Total_Amount"].sum().reset_index())
season_sales
```

**Explanation**

* The transaction month is used to assign a season.
* Each transaction is grouped into a seasonal category.
* compare sales performance across different seasons.

**Justify Your Feature**

* I created this aggregation to identify which season generates the highest sales.

# **Notebook 14 – Domain-Based Feature Engineering**

- Domain-based feature engineering means using business and industry knowledge to create useful features from raw data. 
- These features help the machine learning model understand real-world patterns better.

```python
# Load Datset
import pandas as pd 
import numpy as np 
df = pd.read_csv("manufacturing_production_machine_performance_dataset.csv")
df.info()
```

## **1. Defect Rate**

- **Feature Name :** Defect_Rate
- **Source Column(s) :** Defective_Units, Production_Units
- **Logic :** Defective_Units / Production_Units
- **Business Meaning :** percentage of products that are defective.
- **ML Relevance :** model understand production quality.
- **Leakage Risk :** Low, if calculated from information available during production.
- **Retain/Remove Decision :** Retain

```python
# Defect Rate
df["Defect_Rate"] = round((df["Defective_Units"] / df["Production_Units"].replace(0, np.nan)),2)*100
print(df[["Defective_Units","Production_Units","Defect_Rate"]].head())
```

**Explanation**

- Defect_Rate shows the percentage of defective products.

## **2. Production Efficiency**

- **Feature Name :** Production_Efficiency
- **Source Column(s) :** Production_Units, Operating_Hours
- **Logic :** Production_Units / Operating_Hours
- **Business Meaning :** Number of units produced per operating hour.
- **ML Relevance :** Helps the model measure production performance.
- **Leakage Risk :** Low, if calculated from available production data.
- **Retain/Remove Decision :** Retain

```python
# Production Efficiency
df["Production_Efficiency"] = (df["Production_Units"] / df["Operating_Hours"].replace(0, np.nan))
print(df[["Production_Units","Operating_Hours","Production_Efficiency"]].head())
```

**Explanation**

- Production_Efficiency shows how many units are produced per hour.

## **3. Machine Utilization**

- **Feature Name :** Machine_Utilization
- **Source Column(s) :** Operating_Hours, Idle_Hours
- **Logic :** Operating_Hours / (Operating_Hours + Idle_Hours)
- **Business Meaning :** Percentage of time the machine is actively used.
- **ML Relevance :** Helps the model understand machine usage and productivity.
- **Leakage Risk :** Low, if calculated from available machine data.
- **Retain/Remove Decision :** Retain

```python
# Machine Utilization
df["Machine_Utilization"] = (df["Operating_Hours"] /(df["Operating_Hours"] + df["Idle_Hours"]))
print(df[["Operating_Hours","Idle_Hours","Machine_Utilization"]].head())
```

**Explanation**

- Machine_Utilization shows how much of the available time the machine is working.

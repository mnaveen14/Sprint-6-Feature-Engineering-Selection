## **4. Create Numerical Features**

### **Understand the Concept**

Numerical features are created from existing numerical columns.  

### **Demonstrate the Concept**

- Create useful numerical features.
- Use existing columns for calculations.
- Check the new features.

```python
# Create numerical features
df["Income_Per_Year_At_Company"] = df["Monthly_Income"] / (df["Years_At_Company"] + 1)
df["Workload_Per_Project"] = df["Working_Hours"] / (df["Projects_Completed"] + 1)
df["Training_Performance"] = df["Training_Hours"] * df["Performance_Rating"]

# new features
print(df[["Income_Per_Year_At_Company","Workload_Per_Project","Training_Performance"]].head())
```

## **Explanation**

- Income_Per_Year_At_Company - compares income with company experience.
- Workload_Per_Project - shows working hours per project.
- Training_Performance - combines training and performance.

## **5. Create Categorical Features**

### **Understand the Concept**

Categorical features group data into meaningful categories.  

### **Demonstrate the Concept**

- Create useful categories from existing columns.
- Convert numerical values into groups.
- Check the new categorical features.

```python
# Create categorical features
df["Age_Group"] = pd.cut(df["Age"],bins=[0, 25, 35, 45, 100],labels=["Young", "Adult", "Middle_Age", "Senior"])
df["Experience_Level"] = pd.cut(df["Years_At_Company"],bins=[-1, 2, 5, 10, 100],labels=["New", "Junior", "Experienced", "Senior"])

# new features
print(df[["Age_Group", "Experience_Level"]].head())
```

## **Explanation**

- Age_Group - groups employees based on age.
- Experience_Level - groups employees based on experience.
- pd.cut() - creates categories from numerical values.

## **6. Create Date/Time Features**

### **Understand the Concept**

Date and time columns can provide useful information for the model.  
We can extract details like year, month, and day from the date.

### **Demonstrate the Concept**

- Convert Joining_Date into date format.
- Extract year and month.
- Calculate employee tenure.

```python
# Joining_Date to datetime format
df["Joining_Date"] = pd.to_datetime(df["Joining_Date"])
```

```python
# Create date features
df["Joining_Year"] = df["Joining_Date"].dt.year
df["Joining_Month"] = df["Joining_Date"].dt.month
df["Joining_Day"] = df["Joining_Date"].dt.day
```

```python
# Create tenure 
df["Tenure_Years"] = ((pd.Timestamp("2026-01-01") - df["Joining_Date"]).dt.days / 365)
```

```python
# new features
print(df[["Joining_Date","Joining_Year","Joining_Month","Joining_Day","Tenure_Years"]].head())
```

## **Explanation**

- Joining_Year - shows the joining year.
- Joining_Month - shows the joining month.
- Joining_Day - shows the joining day.
- Tenure_Years - shows how long the employee has worked.

## **7. Create Aggregation Features**

### **Understand the Concept**

Aggregation features combine values to get useful summary information.  

### **Demonstrate the Concept**

- Calculate average values by department.
- Calculate average salary by job level.
- Add the aggregated values as new features.

```python
# Create Aggregation Features
df["Department_Avg_Salary"] = df.groupby("Department")["Monthly_Income"].transform("mean")
df["Job_Level_Avg_Performance"] = df.groupby("Job_Level")["Performance_Rating"].transform("mean")

# new features
print(df[["Department","Department_Avg_Salary","Job_Level","Job_Level_Avg_Performance"]].head())
```

## **Explanation**

- Department_Avg_Salary - shows the average salary in the department.
- Job_Level_Avg_Performance - shows the average performance for the job level.
- groupby() - groups the data for aggregation.
- transform() - adds the result back to each row.

## **8. Create Interaction Features**

### **Understand the Concept**

Interaction features combine two or more features.  

### **Demonstrate the Concept**

- Combine salary and performance.
- Combine working hours and overtime hours.

```python
# Create interaction features
df["Salary_Performance"] = (df["Monthly_Income"] * df["Performance_Rating"])
df["Total_Work_Hours"] = (df["Working_Hours"] + df["Overtime_Hours"])
df["Training_Per_Project"] = (df["Training_Hours"] / (df["Projects_Completed"] + 1))

# new features
print(df[["Salary_Performance","Total_Work_Hours","Training_Per_Project"]].head())
```

## **Explanation**

- Salary_Performance - combines salary and performance.
- Total_Work_Hours - combines working and overtime hours.
- Training_Per_Project - shows training hours for each project.

## **9. Perform Feature Selection**

### **Understand the Concept**

Feature selection means choosing the most useful features for the model and remove unnecessary features.

### **Demonstrate the Concept**

- Separate features and target.
- Check the numerical features.
- Select features that are useful for prediction.

```python
from sklearn.feature_selection import SelectKBest, mutual_info_classif

# features and target
X = df.drop("Attrition", axis=1)
y = df["Attrition"]
```

```python
# numerical features
X_numeric = X.select_dtypes(include="number").fillna(0)
```

```python
# 5 features
selector = SelectKBest(score_func=mutual_info_classif, k=5)
X_selected = selector.fit_transform(X_numeric, y)
```

```python
# selected features
selected_features = X_numeric.columns[selector.get_support()]
print("Selected Features:")
print(selected_features.tolist())
```

## **Explanation**

- X - contains the input features.
- y - contains the target.
- SelectKBest() - selects the best features.
- mutual_info_classif() - checks feature relationship with the target.
- get_support() - returns the selected features.

## **10. Analyze Feature Importance**

### **Understand the Concept**

Feature importance shows which features are more useful for making predictions.  
Understand which features have more impact on the model.

### **Demonstrate the Concept**

- Train a Random Forest model.
- Calculate feature importance.
- Sort the features by importance.

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(n_estimators=100,random_state=42)
model.fit(X_numeric, y)
```

```python
# feature importance
importance = pd.DataFrame({"Feature": X_numeric.columns,"Importance": model.feature_importances_})
importance = importance.sort_values(by="Importance",ascending=False)
print(importance)
```

## **Explanation**

- RandomForestClassifier() - creates the model.
- feature_importances_ - gives importance scores.
- sort_values() - sorts features by importance.

## **11. Check for Feature Leakage**

### **Understand the Concept**

Feature leakage happens when a feature contains information that would not be available when making the prediction.  

### **Demonstrate the Concept**

- Check features related to the target.
- Look for features created using future information.
- Identify features that may cause leakage.

```python
# correlation with target
df["Attrition"] = df["Attrition"].map({"Yes": 1,"No": 0})
correlation = df.select_dtypes(include="number").corr()["Attrition"]
print("Feature Correlation with Attrition:")
print(correlation.sort_values(ascending=False))
```

## **Explanation**

- map() - converts Attrition into 0 and 1.
- corr() - checks the relationship between features.
- High correlation can help identify possible leakage.

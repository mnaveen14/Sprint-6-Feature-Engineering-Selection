# **Notebook 5 – Aggregation Features**

```python
import pandas as pd 
df = pd.read_csv("hospital_patient_treatment_dataset.csv")
df.info()
```

## **1. Count**

**Understand the Concept**

* Count means finding how many records are present in a group.
* It helps us understand the number of patients, visits, or treatments.

**Demonstrate the Concept**

**Example:**

* Count the number of visits for each department.

```python
department_visit_count = (df.groupby("Department")["Visit_ID"].count().reset_index(name="Visit_Count"))
department_visit_count
```

**Explanation**

* groupby() groups the data by Department.
* count() counts the number of visits in each department.

**Justify Your Feature**

* I created this feature to understand the number of patient visits in each department.

## **7. Standard Deviation**

**Understand the Concept**

* Standard deviation measures how much values vary from the average.
* A low value means the values are closer together.

**Demonstrate the Concept**

**Example:**

## **8. Group-Based Aggregation**

**Understand the Concept**

* Group-based aggregation calculates multiple statistics for different groups.
* It allows us to summarize data using more than one measure.

**Demonstrate the Concept**

**Example:**

* Calculate the number of visits, total treatment cost, and average treatment cost for each department.

```python
department_summary = (df.groupby("Department").agg(Visit_Count=("Visit_ID", "count"),
                                                   Total_Treatment_Cost=("Treatment_Cost", "sum"),
                                                   Average_Treatment_Cost=("Treatment_Cost", "mean")).reset_index())
department_summary
```

**Explanation**

* groupby() creates groups based on Department.
* agg() applies multiple aggregation functions.

**Justify Your Feature**

* I created these features to get a complete summary of department-level patient activity.

## **9. Patient-Level Aggregation**

**Understand the Concept**

* Patient-level aggregation summarizes information for each patient.
* It combines multiple visits of the same patient into useful values.

**Demonstrate the Concept**

**Example:**

* Calculate each patient's total treatment cost, visit count, and average treatment cost.

```python
patient_summary = (df.groupby("Patient_ID")  .agg(Visit_Count=("Visit_ID", "count"),
                                                  Total_Treatment_Cost=("Treatment_Cost", "sum"),
                                                  Average_Treatment_Cost=("Treatment_Cost", "mean")).reset_index())
patient_summary.head()
```

**Explanation**

* groupby() groups all visits belonging to the same Patient_ID.
* count() calculates the number of visits.
* sum() calculates total treatment cost.
* mean() calculates average treatment cost.

**Justify Your Feature**

* I created these features to understand each patient's overall treatment history.

## **10. Treatment-Level Aggregation**

**Understand the Concept**

* Treatment-level aggregation summarizes information for each treatment type.
* It helps compare different treatments.

**Demonstrate the Concept**

**Example:**

* Calculate the number of patients and average treatment cost for each treatment type.

```python
treatment_summary = (df.groupby("Treatment_Type").agg( Visit_Count=("Visit_ID", "count"),
                                                       Total_Treatment_Cost=("Treatment_Cost", "sum"),
                                                       Average_Treatment_Cost=("Treatment_Cost", "mean")).reset_index())
treatment_summary
```

**Explanation**

* groupby() groups the data by Treatment_Type.
* count() calculates the number of visits for each treatment.
* sum() calculates the total treatment cost.
* mean() calculates the average treatment cost.

**Justify Your Feature**

* I created these features to compare treatment usage and treatment costs.

## **11. Transaction-Level Aggregation**

**Understand the Concept**

* Transaction-level aggregation summarizes information for each hospital visit.
* It combines related values to create a useful visit-level feature.

**Demonstrate the Concept**

**Example:**

* Calculate the total activity count for each visit using medication and lab test counts.

```python
df["Total_Clinical_Activities"] = (df["Medication_Count"] +  df["Lab_Test_Count"])
df[["Visit_ID","Medication_Count","Lab_Test_Count","Total_Clinical_Activities"]].head()
```

**Explanation**

* The medication count and lab test count are added together.
* The result is stored as Total_Clinical_Activities.
* It represents the total number of recorded medication and lab-test activities for a visit.

**Justify Your Feature**

* I created this feature to measure the overall clinical activity during each patient visit.

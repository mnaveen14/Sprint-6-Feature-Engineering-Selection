# **Notebook 9 – Feature Selection**

```python
# Load Datset
import pandas as pd
df = pd.read_csv("telecom_customer_churn.csv")
df.info()
```

## **1. What is Feature Selection?**

**Understand the Concept**

* Feature Selection means choosing the most useful features from the dataset.
* It removes features that are not helpful.
* It helps us work with a smaller and better set of features.

## **2. Why Feature Selection?**

**Understand the Concept**

* Feature selection helps us keep only useful features.
* Removes unnecessary features.
* Reduces the complexity of the model.
* Makes model training faster.
* Helps reduce overfitting.
* Can improve model performance.

## **3. Relevant Features**

**Understand the Concept**

* Relevant features are features that provide useful information about the target.
* They can help the model make better predictions.

**Demonstrate the Concept**

* Tenure_Months, Monthly_Charges, and Support_Calls may help predict Churn.

```python
relevant_features = df[["Tenure_Months", "Monthly_Charges", "Support_Calls", "Churn"]]
relevant_features.head()
```

**Explanation**

* Tenure_Months, Monthly_Charges, and Support_Calls are selected.

  
**Justify Your Feature**

* I selected these features because they may help explain why a customer churns.

## **4. Irrelevant Features**

**Understand the Concept**

* Irrelevant features do not help predict the target.
* They can be removed from the dataset.

**Demonstrate the Concept**

* Customer_ID only identifies a customer.
* It does not help predict Churn.

```python
df["Customer_ID"].head()
```

**Explanation**

* Customer_ID is only used to identify customers.
* It does not provide useful information about Churn.

**Justify Your Feature**

* I identified Customer_ID as an irrelevant feature because it does not help predict Churn.

## **5. Redundant Features**

**Understand the Concept**

* Redundant features are features that give the same or very similar information.
* They may not add much new information to the model.

**Demonstrate the Concept**

* Monthly_Charges and Monthly_Charges_Copy contain the same information.
* We can remove one of them.

```python
df[["Monthly_Charges", "Monthly_Charges_Copy"]].head()
```

**Explanation**

* Both features contain the same values.
* Keeping both features is unnecessary.
* We can keep one and remove the other.

**Justify Your Feature**

* I identified Monthly_Charges_Copy as redundant because it duplicates Monthly_Charges.

## **6. Filter Methods**

**Understand the Concept**

* Filter methods select features using statistical tests.
* They check which features are useful before training the model.

**Demonstrate the Concept**

* We can use correlation to find features related to the target.
* Features with useful relationships can be selected.
  

```python
# correlation between numerical features
numeric_features = df.select_dtypes(include="number")
correlation = numeric_features.corr()
correlation
```

**Explanation**

* Correlation shows how features are related.
* It helps find useful features.

**Justify Your Feature**

* I used a filter method to quickly identify potentially useful features.

## **7. Wrapper Methods**

**Understand the Concept**

* Wrapper methods select features by testing them with a model.
* They try different feature combinations to find the best ones.

**Demonstrate the Concept**

* RFE is a wrapper method.
* It removes less useful features step by step.

```python
from sklearn.feature_selection import RFE
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer

X = df.select_dtypes(include="number").drop(columns=["Constant_Feature"])
X = SimpleImputer(strategy="median").fit_transform(X)

y = df["Churn"].map({"No": 0, "Yes": 1})

model = RandomForestClassifier(n_estimators=50, random_state=42)

rfe = RFE(model, n_features_to_select=5)
rfe.fit(X, y)

print(rfe.support_)
```

**Explanation**

* RFE removes less useful features one by one.
* It keeps the selected features.

**Justify Your Feature**

* I used RFE to select the best features for the model.

## **8. Embedded Methods**

**Understand the Concept**

* Embedded methods select features while the model is being trained.
* They identify important features automatically.

**Demonstrate the Concept**

* Lasso and tree-based models can select important features during training.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer

X = df.select_dtypes(include="number").drop(columns=["Constant_Feature"])
X = SimpleImputer(strategy="median").fit_transform(X)

y = df["Churn"].map({"No": 0, "Yes": 1})

model = RandomForestClassifier(n_estimators=50, random_state=42)
model.fit(X, y)

print(model.feature_importances_)
```

**Explanation**

* The model gives an importance score to each feature.
* Features with higher scores are more important.

**Justify Your Feature**

* I used an embedded method to find important features during model training.

## **9. Variance Threshold**

**Understand the Concept**

* Variance shows how much a feature changes.
* Features with very little change may not be useful.

**Demonstrate the Concept**

* Constant_Feature has the same value in every row.
* It has no useful variation, so we can remove it.

```python
from sklearn.feature_selection import VarianceThreshold

X = df.select_dtypes(include="number")

selector = VarianceThreshold(threshold=0)
X_selected = selector.fit_transform(X)

print(X_selected.shape)
```

**Explanation**

* The selector removes features with no variation.
* Constant_Feature is removed.

**Justify Your Feature**

* I used Variance Threshold to remove features with no variation.

## **10. Correlation-Based Selection**

**Understand the Concept**

* Correlation shows how two numerical values are related.
* It can help find features related to the target.

```python
df["Churn_Numeric"] = df["Churn"].map({"No": 0, "Yes": 1})

correlation = df.select_dtypes(include="number").corr()["Churn_Numeric"]

print(correlation.sort_values(ascending=False))
```

**Explanation**

* Correlation values show the relationship with Churn.
* Stronger relationships may indicate useful features.

**Justify Your Feature**

* I used correlation to find features related to Churn.

## **11. Mutual Information**

**Understand the Concept**

* Mutual Information shows how much information a feature gives about the target.
* A higher score means the feature may be more useful.

```python
from sklearn.feature_selection import mutual_info_classif
from sklearn.impute import SimpleImputer

X = df.select_dtypes(include="number").drop(columns=["Churn_Numeric"])
X = SimpleImputer(strategy="median").fit_transform(X)

y = df["Churn_Numeric"]

scores = mutual_info_classif(X, y, random_state=42)

print(scores)
```

**Explanation**

* Each feature gets an information score.
* Higher scores show more useful features.

**Justify Your Feature**

* I used Mutual Information to find features that provide useful information about Churn.

## **12. Chi-Square**

**Understand the Concept**

* Chi-Square checks the relationship between categorical features and the target.
* It helps find useful categorical features.

```python
from sklearn.feature_selection import chi2
from sklearn.preprocessing import LabelEncoder

X = df[["Gender", "Plan_Type", "Contract_Type", "Payment_Method"]].copy()

for column in X.columns:
    X[column] = LabelEncoder().fit_transform(X[column].astype(str))

y = df["Churn_Numeric"]

scores, p_values = chi2(X, y)

print(scores)
```

**Explanation**

* Chi-Square checks each categorical feature against Churn.
* Higher scores can show a stronger relationship.

**Justify Your Feature**

* I used Chi-Square to find useful categorical features.

## **13. Recursive Feature Elimination (RFE)**

**Understand the Concept**

* RFE removes less useful features step by step.
* It keeps the required number of important features.

```python
from sklearn.feature_selection import RFE
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer

X = df.select_dtypes(include="number").drop(columns=["Churn_Numeric"])
X = SimpleImputer(strategy="median").fit_transform(X)

y = df["Churn_Numeric"]

model = RandomForestClassifier(n_estimators=50, random_state=42)

rfe = RFE(model, n_features_to_select=5)
rfe.fit(X, y)

print(rfe.support_)
```

**Explanation**

* RFE removes less useful features one by one.
* It keeps the selected features.

**Justify Your Feature**

* I used RFE to select the best features for the model.

## **14. L1/Lasso-Based Selection**

**Understand the Concept**

* L1 regularization can reduce the importance of less useful features.
* Features with zero coefficients can be removed.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

X = df.select_dtypes(include="number").drop(columns=["Churn_Numeric"])
X = SimpleImputer(strategy="median").fit_transform(X)
X = StandardScaler().fit_transform(X)

y = df["Churn_Numeric"]

model = LogisticRegression(
    penalty="l1",
    solver="liblinear",
    random_state=42
)

model.fit(X, y)

print(model.coef_)
```

**Explanation**

* L1 can make some feature coefficients zero.
* Features with zero coefficients can be removed.

**Justify Your Feature**

* I used L1 to identify features that have little contribution to the model.

## **15. Tree-Based Feature Importance**

**Understand the Concept**

* Tree models can show which features are important.
* Higher importance means the feature has more influence on the model.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.impute import SimpleImputer

X = df.select_dtypes(include="number").drop(columns=["Churn_Numeric"])
X = SimpleImputer(strategy="median").fit_transform(X)

y = df["Churn_Numeric"]

model = RandomForestClassifier(n_estimators=50, random_state=42)
model.fit(X, y)

importance = pd.Series(
    model.feature_importances_,
    index=df.select_dtypes(include="number").drop(columns=["Churn_Numeric"]).columns
)

print(importance.sort_values(ascending=False))
```

**Explanation**

* The tree model gives an importance score to each feature.
* Higher scores show more important features.

**Justify Your Feature**

* I used tree-based importance to identify the most useful features for Churn prediction.

## **Advantages and Limitations**

* **Variance Threshold:** Simple and fast, but may remove useful low-variation features.
* **Correlation:** Easy to use, but mainly finds linear relationships.
* **Mutual Information:** Can find different types of relationships, but scores can be harder to understand.
* **Chi-Square:** Good for categorical features, but needs suitable data.
* **RFE:** Can find a good feature set, but can take more time.
* **L1/Lasso:** Can remove less useful features automatically, but depends on the regularization setting.
* **Tree-Based Importance:** Finds important features well, but importance scores can sometimes be misleading.

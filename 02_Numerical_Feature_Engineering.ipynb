# **Notebook 2 – Numerical Feature Engineering**

```python
# Load dataset
import pandas as pd
df = pd.read_csv("online_retail_sales_dataset.csv")
df.info()
```

## **1. Mathematical Transformations**

**Understand the Concept**

* Mathematical transformation means using existing numbers to create new information.
* We can use simple operations like `+`, `-`, `×`, and `÷`.
* These operations help create useful numerical features.

## **2. Addition**

**Understand the Concept**

* Addition combines two or more numerical values.
* It can create a new feature that represents a total value.
* Use addition when different values contribute to the same measurement.

**Demonstrate the Concept**

**Example:**

* Add `Net_Sales` and `Shipping_Cost` to find the total amount including shipping.

**AI/ML Usage:**

* Addition can create total-value features that give the model more useful information.

```python
# addition
df["Total_Amount_With_Shipping"] = (df["Net_Sales"] + df["Shipping_Cost"])
df[["Net_Sales", "Shipping_Cost", "Total_Amount_With_Shipping"]].head()
```

## **3. Subtraction**

**Understand the Concept**

* Subtraction finds the difference between two numerical values.
* It helps us understand the gap between two amounts.
* Use subtraction when one value needs to be compared with another.

**Demonstrate the Concept**

**Example:**

* Subtract `Cost_Price` from `Unit_Price` to find the profit amount per item before other costs.

**AI/ML Usage:**

* Subtraction can create features that show differences or gaps between important values.

```python
# subtraction

df["Price_Margin_Per_Item"] = (df["Unit_Price"] - df["Cost_Price"])
df[["Unit_Price", "Cost_Price", "Price_Margin_Per_Item"]].head()
```

**Justify Your Feature**

* I created this feature because it shows the difference between selling price and cost price.

## **4. Division**

**Understand the Concept**

* Division compares one value with another.
* It can help calculate an average or value per unit.
* Use division when one value depends on the size of another value.

**Demonstrate the Concept**

*m*Example:**

* Divide `Net_Sales` by `Quantity` to find the sales amount per item.

**AI/ML Usage:**

* Division can create per-unit features that give the model more useful information.

```python
# division 
df["Sales_Per_Item"] = (df["Net_Sales"] / df["Quantity"])
df[["Net_Sales", "Quantity", "Sales_Per_Item"]].head()
```

## **5. Ratios**

**Understand the Concept**

* A ratio compares two numerical values.
* It shows how one value relates to another.
* Ratios are useful when the relationship between two values is important.

**Demonstrate the Concept**

**Example:**

* Compare `Profit` with `Net_Sales` to understand profit relative to sales.

**AI/ML Usage:**

* Ratio features help the model understand relationships between numerical values.

```python
# ratio
df["Profit_to_Sales_Ratio"] = (df["Profit"] / df["Net_Sales"])
df[    ["Profit", "Net_Sales", "Profit_to_Sales_Ratio"]].head()
```

## **6. Percentages**

**Understand the Concept**

* A percentage shows a value as part of another value.
* It makes comparisons easier to understand.
* Percentages are useful for measuring discounts, costs, or contributions.

**Demonstrate the Concept**

**Example:**

* Calculate shipping cost as a percentage of `Net_Sales`.

**AI/ML Usage:**

* Percentage features help the model understand the relative contribution of different values.

```python
# Calculate shipping cost percentage

df["Shipping_Cost_Percent"] = (df["Shipping_Cost"] / df["Net_Sales"]) * 100
df[["Shipping_Cost", "Net_Sales", "Shipping_Cost_Percent"]].head()
```

## **7. Differences**

**Understand the Concept**

* Difference measures the gap between two numerical values.
* It helps us compare actual values with another reference value.
* Differences can show increases, decreases, or gaps.

**Demonstrate the Concept**

**Example:**

* Compare `Unit_Price` with `Cost_Price` to find the difference between selling and cost price.

**AI/ML Usage:**

* Difference features help the model understand gaps between related numerical values.

```python
# Calculate the difference between selling price and cost price
df["Price_Difference"] = (df["Unit_Price"] - df["Cost_Price"])
df[["Unit_Price", "Cost_Price", "Price_Difference"]].head()
```

## **8. Aggregations**

**Understand the Concept**

* Aggregation combines multiple values into a summary value.
* Common aggregations include sum, mean, minimum, and maximum.
* Aggregations help summarize numerical information.

**Demonstrate the Concept**

**Example:**

* Calculate the average of `Unit_Price`, `Cost_Price`, and `Shipping_Cost` for each transaction.

**AI/ML Usage:**

* Aggregated features summarize information and can provide useful inputs to a model.

```python
# Calculate the average value of selected numerical columns

df["Average_Cost_Measure"] = df[["Unit_Price", "Cost_Price", "Shipping_Cost"]].mean(axis=1)
df[["Unit_Price", "Cost_Price", "Shipping_Cost","Average_Cost_Measure"]].head()
```


**Justify Your Feature**

* I created this feature because it summarizes multiple cost-related values into one feature.

## **9. Log Transformation**

**Understand the Concept**

* Log transformation changes large numerical values into a smaller scale.
* It is useful when data is highly skewed.
* It can reduce the effect of extremely large values.

**Demonstrate the Concept**

**Example:**

* `Gross_Sales` may contain some very large transaction values, so a log transformation can reduce their effect.

**AI/ML Usage:**

* Log transformation can help some Machine Learning models work better with highly skewed numerical features.

```python
import numpy as np
# log transformation
df["Log_Gross_Sales"] = np.log1p(df["Gross_Sales"])
df[["Gross_Sales", "Log_Gross_Sales"]].head()
```

**Justify Your Feature**

* I created this feature because it helps reduce the effect of highly large sales values.

## **10. Square Root Transformation**

**Understand the Concept**

* Square root transformation reduces the size of large numerical values.
* It is useful for reducing moderate skewness.
* It is less aggressive than log transformation.

**Demonstrate the Concept**

**Example:**

* Apply a square root transformation to `Quantity`.

**AI/ML Usage:**

* Square root transformation can make a numerical feature less skewed and easier for some models to use.

```python
# square root transformation
df["Sqrt_Quantity"] = np.sqrt(df["Quantity"])
df[["Quantity", "Sqrt_Quantity"]].head()
```

## **11. Power Transformation**

**Understand the Concept**

* Power transformation changes the scale of a numerical feature using a power.
* It can help change the shape of a feature's distribution.
* The power used depends on the data and the purpose of the transformation.

**Demonstrate the Concept**

**Example:**

* Apply a square power to `Product_Rating` to create a new transformed feature.

**AI/ML Usage:**

* Power transformations can help create features with a different scale or distribution.

```python
# power transformation
df["Rating_Squared"] = df["Product_Rating"] ** 2
df[["Product_Rating", "Rating_Squared"]].head()
```

## **12. Absolute Difference**

**Understand the Concept**

* Absolute difference measures the size of a difference.
* It always gives a positive value.
* It is useful when the direction of the difference is not important.

**Demonstrate the Concept**

**Example:**

* Find the absolute difference between `Unit_Price` and `Cost_Price`.

**AI/ML Usage:**

* Absolute difference helps the model understand the size of a gap between two features.

```python
# absolute price difference
df["Absolute_Price_Difference"] = (df["Unit_Price"] - df["Cost_Price"]).abs()
df[["Unit_Price", "Cost_Price", "Absolute_Price_Difference"]].head()
```

**Explanation**

* The absolute value removes the negative sign from any difference.
* The new feature shows the size of the price gap.

**Justify Your Feature**

* I created this feature because it clearly shows the size of the difference between selling and cost price.

## **13. Relative Difference**

**Understand the Concept**

* Relative difference compares a difference with a reference value.
* It shows how large the difference is relative to that value.
* It is useful when values have different scales.

**Demonstrate the Concept**

**Example:**

* Compare the difference between `Unit_Price` and `Cost_Price` relative to `Cost_Price`.

**AI/ML Usage:**

* Relative difference helps the model understand the size of a difference in relative terms.

```python
# Calculate relative price difference

df["Relative_Price_Difference"] = ((df["Unit_Price"] - df["Cost_Price"])/ df["Cost_Price"])
df[["Unit_Price", "Cost_Price", "Relative_Price_Difference"]].head()
```

**Explanation**

* The price difference is divided by the cost price.
* The new feature shows the price difference relative to the cost.

**Justify Your Feature**

* I created this feature because it shows the price difference in relative terms.

## **14. Rate Features**

**Understand the Concept**

* A rate feature measures an amount relative to another quantity.
* It can show activity per item, day, or other unit.
* Rates help compare transactions of different sizes.

**Demonstrate the Concept**

**Example:**

* Calculate sales per delivery day using `Net_Sales` and `Delivery_Days`.

**AI/ML Usage:**

* Rate features help the model compare performance across different quantities or time periods.

```python
# Calculate sales per delivery day

df["Sales_Per_Delivery_Day"] = (df["Net_Sales"] / df["Delivery_Days"])
df[["Net_Sales", "Delivery_Days", "Sales_Per_Delivery_Day"]].head()
```


**Justify Your Feature**

* I created this feature because it compares sales value with delivery time.

## **15. Normalized Features**

**Understand the Concept**

* Normalization changes numerical values to a common scale.
* It helps features with different ranges become easier to compare.
* Min-Max normalization usually converts values to a range between 0 and 1.

**Demonstrate the Concept**

**Example:**

* `Gross_Sales` can have a much larger range than `Product_Rating`.
* We can normalize `Gross_Sales` to a 0–1 range.

**AI/ML Usage:**

* Normalized features are useful for algorithms that are sensitive to feature scale, such as KNN and some gradient-based models.

```python
# Normalize - Gross_Sales using Min-Max scaling

from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()

df["Normalized_Gross_Sales"] = scaler.fit_transform(df[["Gross_Sales"]])
df[["Gross_Sales", "Normalized_Gross_Sales"]].head()
```

**Explanation**

* `MinMaxScaler` converts `Gross_Sales` to a range between 0 and 1.
* The original values are kept unchanged.
* The normalized feature makes the values easier to compare with other scaled features.

**Justify Your Feature**

* I created this feature because it puts sales values on a common scale for Machine Learning.

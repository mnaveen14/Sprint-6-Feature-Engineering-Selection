## **Section 1: Concept Questions**



### **Q1. What is the main purpose of Feature Engineering?**

**Answer:** Feature Engineering is the process of creating or changing features from existing data to give the machine learning model more useful information.


### **Q2. What is the difference between a numerical feature and a categorical feature?**

**Answer:**

- **Numerical Feature** - contains numbers, such as Age or Income.
- **Categorical Feature** - contains groups or categories, such as Department or Gender.

### **Q3. What is the difference between nominal and ordinal variables?**

**Answer:**

- **Nominal** - categories have no order, such as City or Department.
- **Ordinal** - categories have a meaningful order, such as Low, Medium, and High.


### **Q4. Why are date and time features useful in machine learning?**

**Answer:** Date and time features can show useful patterns such as seasonality, trends, weekdays, months, and time periods.


### **Q5. What are aggregation features? Give two examples.**

**Answer:** Aggregation features summarize multiple records into useful values.

Examples:
- Total spending per customer.
- Average transaction amount per customer.


### **Q6. What are interaction features? Why are they useful?**

**Answer:** Interaction features combine two or more features to capture their relationship.

Example: Income / Dependents can show income available per dependent.


### **Q7. What is the difference between equal-width and equal-frequency binning?**

**Answer:**

- **Equal-Width Binning** - divides the value range into bins with the same size.
- **Equal-Frequency Binning** - creates bins containing nearly the same number of records.



### **Q8. What is rare category grouping and why is it used?**

**Answer:** Rare category grouping combines categories that appear very few times into one category. It helps reduce unnecessary categories and makes the data easier to handle.


### **Q9. What is the difference between Bag of Words and TF-IDF?**

**Answer:**

- **Bag of Words** - represents text using word counts.
- **TF-IDF** - gives more importance to words that are common in a document but less common across all documents.


### **Q10. What is the difference between filter, wrapper, and embedded feature selection methods?**

**Answer:**

- **Filter** - selects features using statistical measures.
- **Wrapper** - selects features by testing different feature combinations with a model.
- **Embedded** - selects important features during model training.


### **Q11. Why should features be scaled before applying PCA?**

**Answer:** PCA is affected by feature scale. Scaling makes sure features with larger values do not dominate the principal components.


### **Q12. What does explained variance mean in PCA?**

**Answer:** Explained variance shows how much information from the original data is captured by each principal component.


### **Q13. Why should feature importance not be treated as proof of causation?**

**Answer:** Feature importance shows how useful a feature is for prediction. It does not prove that the feature directly causes the target outcome.


### **Q14. Why should feature engineering be performed carefully to avoid data leakage?**

**Answer:** Data leakage can happen when a feature uses information that would not be available at prediction time. This can make the model performance look unrealistically high.

### **Q15. Numerical Feature Engineering**

Given Income, LoanAmoun, MonthlyExpenses, and Savings, create at least four useful numerical features.


```python
import pandas as pd

df = pd.DataFrame({"Income": [40000, 60000, 80000],
                   "LoanAmount": [200000, 300000, 500000],
                   "MonthlyExpenses": [15000, 20000, 30000],
                   "Savings": [50000, 100000, 150000]})

df["Loan_to_Income"] = df["LoanAmount"] / (df["Income"] * 12)
df["Expense_to_Income"] = df["MonthlyExpenses"] / (df["Income"] / 12)
df["Savings_to_Income"] = df["Savings"] / df["Income"]
df["Income_After_Expenses"] = (df["Income"] / 12) - df["MonthlyExpenses"]

df
```

### **Q16. Categorical Feature Engineering**

Create a categorical feature from an existing categorical or numerical column using a suitable encoding technique.

```python
df = pd.DataFrame({"Education": ["Graduate", "Postgraduate", "High School", "Graduate"]})
df["Education_Code"] = df["Education"].map({"High School": 1,"Graduate": 2,"Postgraduate": 3})
df
```

### **Q17. Date and Time Feature Engineering**

From the ApplicationDate column, create:
- Year
- Month
- Day
- Day of Week
- Weekend Indicator


```python
df = pd.DataFrame({"ApplicationDate": pd.to_datetime(["2025-01-15","2025-06-20","2025-12-05"])})

df["Year"] = df["ApplicationDate"].dt.year
df["Month"] = df["ApplicationDate"].dt.month
df["Day"] = df["ApplicationDate"].dt.day
df["DayOfWeek"] = df["ApplicationDate"].dt.dayofweek
df["IsWeekend"] = df["DayOfWeek"] >= 5

df
```

### **Q18. Aggregation Features**

Create customer-level aggregation features such as:
- Total spending
- Average spending
- Transaction count
- Maximum spending

```python
df = pd.DataFrame({"CustomerID": [1, 1, 2, 2, 2],"Amount": [500, 800, 300, 700, 400]})

customer_agg = df.groupby("CustomerID").agg(Total_Spending=("Amount", "sum"),
                                            Average_Spending=("Amount", "mean"),
                                            Transaction_Count=("Amount", "count"),
                                            Maximum_Spending=("Amount", "max")).reset_index()

customer_agg
```

### **Q19. Interaction Features**

Create at least three interaction features using Age, Income, LoanAmount, and CreditScore.



```python
df = pd.DataFrame({"Age": [25, 35, 45],"Income": [40000, 60000, 80000],
                   "LoanAmount": [200000, 300000, 500000],"CreditScore": [650, 720, 780]})

df["Income_Age"] = df["Income"] * df["Age"]
df["Loan_to_Income"] = df["LoanAmount"] / (df["Income"] + 1)
df["Income_CreditScore"] = df["Income"] * df["CreditScore"]

df
```

### **Q20. Binning and Discretization**

Divide the Age column into meaningful groups using binning.

```python
df = pd.DataFrame({"Age": [22, 28, 35, 42, 55, 65]})
df["Age_Group"] = pd.cut(df["Age"],bins=[0, 25, 35, 50, 100],labels=["Young", "Adult", "Middle_Age", "Senior"])
df
```

### **Q21. Text Feature Engineering**

From a Review column, create:
- Text Length
- Word Count
- Digit Count
- Special Character Count


```python
df = pd.DataFrame({"Review": ["Good product!","Very useful product 10/10","Not good, damaged item."]})
df["Text_Length"] = df["Review"].str.len()
df["Word_Count"] = df["Review"].str.split().str.len()
df["Digit_Count"] = df["Review"].str.count(r"\d")
df["Special_Character_Count"] = df["Review"].str.count(r"[^a-zA-Z0-9\s]")

df
```

### **Q23. Feature Importance**

Train a Random Forest model and find the top five important features.


```python
from sklearn.ensemble import RandomForestClassifier

X = pd.DataFrame({"Age": [22, 30, 40, 50, 60, 35],
                  "Income": [30000, 50000, 70000, 90000, 40000, 60000],
                  "CreditScore": [600, 680, 750, 800, 620, 720]})
y = [1, 0, 0, 0, 1, 0]

model = RandomForestClassifier(n_estimators=100,random_state=42)
model.fit(X, y)
importance = pd.Series(model.feature_importances_,index=X.columns).sort_values(ascending=False)
print(importance.head(5))
```

### **Q24. PCA**

Apply PCA to numerical features and reduce them to two principal components. Check the explained variance.



```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Scale features
X_scaled = StandardScaler().fit_transform(X)

# Apply PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

print("Explained Variance:")
print(pca.explained_variance_ratio_)
```

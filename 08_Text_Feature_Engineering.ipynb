# **Notebook 8 – Text Feature Engineering**

Text Feature Engineering means converting raw text into useful numerical features that a Machine Learning model can understand.

```python
# Load Datset
import pandas as pd
df = pd.read_csv("ecommerce_customer_reviews_dataset.csv")
df.info()
```

## **1. Text Length**

**Understand the Concept**

* Text Length means finding the total no of characters in a text.
* It helps us understand how long a review is.

**Demonstrate the Concept**

**Example:**

* "Good product" has a shorter length than "This is an excellent product and I really like it."

```python
# Text Length = characters + spaces
df["Text_Length"] = df["Review_Text"].fillna("").str.len()
df[["Review_Text", "Text_Length"]].head()
```

**Explanation**

* fillna() replaces missing reviews with an empty text.
* str.len() counts the characters in each review.

**Justify Your Feature**

* I created this feature to measure how long each customer review is.

## **2. Character Count**

**Understand the Concept**

* Character Count means counting the characters in a text.
* It shows how many characters are present in each review.

**Demonstrate the Concept**

**Example:**

* "Good product" contains 12 characters including the space.

```python
# Character Count = characters without spaces
df["Character_Count"] = df["Review_Text"].fillna("").str.replace(" ", "", regex=False).str.len()
df[["Review_Text", "Character_Count"]].head()
```

**Explanation**

* Spaces are removed before counting the characters.
* The number of characters is stored in Character_Count.

**Justify Your Feature**

* I created this feature to measure the number of characters in each review.

## **3. Word Count**

**Understand the Concept**

* Word Count means counting the number of words in a text.
* It helps us understand how much information a review contains.

**Demonstrate the Concept**

**Example:**

* "Good product and fast delivery" has 5 words.

```python
df["Word_Count"] = df["Review_Text"].fillna("").str.split().str.len()
df[["Review_Text", "Word_Count"]].sample(6)
```

**Explanation**

* str.split() separates the review into individual words.
* str.len() counts the words in each review.

**Justify Your Feature**

* I created this feature to measure the number of words in each customer review.

## **4. Sentence Count**

**Understand the Concept**

* Sentence Count means counting the number of sentences in a text.


**Demonstrate the Concept**

**Example:**

* "Good product. Fast delivery!" has 2 sentences.

```python
import re

df["Sentence_Count"] = df["Review_Text"].fillna("").apply(lambda x: len([s for s in re.split(r"[.!?]+", x) if s.strip()]))
df[["Review_Text", "Sentence_Count"]].head()
```

**Explanation**

* re.split() separates the review at sentence endings.
* The sentences are counted and stored in Sentence_Count.

**Justify Your Feature**

* I created this feature to measure the number of sentences in each review.

## **5. Number of Digits**

**Understand the Concept**

* This feature counts the numbers present in a review.
* It can help identify reviews that contain order numbers, prices, ratings, or other numbers.

**Demonstrate the Concept**

**Example:**

* "I ordered 2 items for 500" contains 4 digits.

```python
df["Digit_Count"] = df["Review_Text"].fillna("").apply(lambda x: sum(char.isdigit() for char in x))
df[["Review_Text", "Digit_Count"]].head()
```

**Explanation**

* isdigit() checks whether each character is a number.
* The digits are counted for each review.

**Justify Your Feature**

* I created this feature to identify how many numbers are present in each review.

## **6. Number of Special Characters**

**Understand the Concept**

* Special characters are symbols such as !, @, #, and $.
* Counting them can help understand how customers write their reviews.

**Demonstrate the Concept**

**Example:**

* "Excellent product!!" contains 2 special characters.

```python
import string

df["Special_Character_Count"] = df["Review_Text"].fillna("").apply(lambda x: sum(char in string.punctuation for char in x))
df[["Review_Text", "Special_Character_Count"]].head()
```

**Explanation**

* string.punctuation contains common special characters.
* Each special character in the review is counted.

**Justify Your Feature**

* I created this feature to measure the use of special characters in reviews.

## **7. Uppercase Count**

**Understand the Concept**

* Uppercase Count means counting capital letters in a review.

**Demonstrate the Concept**

**Example:**

* "GOOD PRODUCT" contains many uppercase letters.

```python
df["Uppercase_Count"] = df["Review_Text"].fillna("").apply(lambda x: sum(char.isupper() for char in x))
df[["Review_Text", "Uppercase_Count"]].head()
```

**Explanation**

* isupper() checks whether each character is uppercase.
* The uppercase letters are counted.
  
**Justify Your Feature**

* I created this feature to identify the use of capital letters in reviews.

## **8. Lowercase Count**

**Understand the Concept**

* Lowercase Count means counting small letters in a review.
* It shows how many lowercase letters are used in the text.

**Demonstrate the Concept**

**Example:**

* "good product" contains only lowercase letters.

```python
df["Lowercase_Count"] = df["Review_Text"].fillna("").apply(lambda x: sum(char.islower() for char in x))
df[["Review_Text", "Lowercase_Count"]].head()
```

**Explanation**

* islower() checks whether each character is lowercase.
* The lowercase letters are counted.
  
**Justify Your Feature**

* I created this feature to measure the use of lowercase letters in reviews.

## **9. Keyword Count**

**Understand the Concept**

* Keyword Count means counting how many times a specific word appears.
* It helps identify important words in customer reviews.

**Demonstrate the Concept**

**Example:**

* Count how many times the word "good" appears in each review.

```python
df["Good_Keyword_Count"] = df["Review_Text"].fillna("").str.lower().str.count(r"\bgood\b")
df[["Review_Text", "Good_Keyword_Count"]].head()
```

**Explanation**

* lower() changes the review to lowercase.
* count() counts how many times the keyword good appears.

**Justify Your Feature**

* I created this feature to identify reviews that contain the keyword good.

## **10. Bag of Words**

**Understand the Concept**

* Bag of Words converts text into numbers based on word counts.
* Each important word becomes a feature.
* The model can then use these numbers.

**Demonstrate the Concept**

**Example:**

* "good product" becomes numerical features based on the words good and product.

```python
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer(max_features=20)
bow_features = vectorizer.fit_transform(df["Review_Text"].fillna(""))

bow_df = pd.DataFrame(bow_features.toarray(),columns=vectorizer.get_feature_names_out())
bow_df.head()
```

**Explanation**

* CountVectorizer finds words in the reviews.
* Each word becomes a numerical feature.
* The values show how many times the words appear.

**Justify Your Feature**

* I created Bag of Words features to convert customer reviews into numerical data for ML models.

## **11. TF-IDF**

**Understand the Concept**

* TF-IDF gives importance to words in a text.
* Common words get lower importance.
* Words that are more useful or specific get higher importance.

**Demonstrate the Concept**

**Example:**

* A word that appears in many reviews may get less importance than a word that appears in only a few reviews.

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer(max_features=20)
tfidf_features = tfidf_vectorizer.fit_transform(df["Review_Text"].fillna(""))

tfidf_df = pd.DataFrame(tfidf_features.toarray(),columns=tfidf_vectorizer.get_feature_names_out())
tfidf_df.head()
```

**Explanation**

* TfidfVectorizer converts reviews into numerical values.
* It gives different importance to different words.

**Justify Your Feature**

* I created TF-IDF features to give more importance to useful words in customer reviews.

## **12. N-Grams**

**Understand the Concept**

* N-Grams are groups of words that appear together.
* They help capture short phrases in text.
* Common types are Unigrams, Bigrams, and Trigrams.

**Demonstrate the Concept**

**Example:**

* "very good product" can create phrases such as "very good" and "good product".

```python
from sklearn.feature_extraction.text import CountVectorizer

ngram_vectorizer = CountVectorizer( ngram_range=(1, 2), max_features=20)
ngram_features = ngram_vectorizer.fit_transform(df["Review_Text"].fillna(""))
ngram_df = pd.DataFrame(ngram_features.toarray(),columns=ngram_vectorizer.get_feature_names_out())
ngram_df.head()
```

**Explanation**

* ngram_range=(1, 2) creates single words and two-word phrases.
* The phrases become numerical features.

**Justify Your Feature**

* I created N-Gram features to capture useful word combinations in customer reviews.
